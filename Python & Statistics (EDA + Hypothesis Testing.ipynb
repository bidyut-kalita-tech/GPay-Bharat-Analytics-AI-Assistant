***Python & Statistics (EDA + Hypothesis Testing***

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
# 1. Load Data
df = pd.read_csv("upi_data.csv")

# 2. View first 5 rows
df.head()
```

**Hypothesis Test 1 — Weekday vs Weekend Spending (Two-Sample T-Test)**

```python
# H0: The average transaction amount on weekdays and weekends is equal.
# H1: There is a significant difference between the two.

weekday_spend = df[df['is_weekend'] == 0]['amount (INR)']
weekend_spend = df[df['is_weekend'] == 1]['amount (INR)']

t_stat, p_val = stats.ttest_ind(weekday_spend, weekend_spend, equal_var=False)

print(f"T-Statistic: {t_stat:.4f}")
print(f"P-Value: {p_val:.4e}")

if p_val < 0.05:
    print("Decision: Reject H0 — Significant difference found in spending behavior between weekdays and weekends.")
else:
    print("Decision: Fail to reject H0 — No significant spending difference.")
```

**Hypothesis Test 2 — Age Segment vs Transaction Type (Chi-Square Test)**

```python
contingency_table = pd.crosstab(df['sender_age_group'], df['transaction type'])
chi2, p_val_chi2, dof, _ = stats.chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"P-Value: {p_val_chi2:.4e}")
```
